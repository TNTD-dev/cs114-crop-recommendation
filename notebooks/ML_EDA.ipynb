```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Crop_recommendation.csv')
```

```python
print("--- 5 dòng đầu tiên ---")
print(df.head())
print((" "))
df.info()
```

```python
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(15, 10))
features = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
for i, col in enumerate(features):
    plt.subplot(3, 3, i+1)
    sns.histplot(df[col], kde=True, color='teal')
    plt.title(f'Phân phối của {col}')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='RdYlGn', fmt='.2f')
plt.title("Ma trận tương quan giữa các yếu tố đất và khí hậu")
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.boxplot(x='rainfall', y='label', data=df, palette='Spectral')
plt.title("Nhu cầu lượng mưa theo từng loại cây trồng")
plt.xlabel("Lượng mưa (mm)")
plt.ylabel("Loại cây")
plt.show()
```

```python
plt.figure(figsize=(15, 5))
sns.boxplot(data=df.drop('label', axis=1))
plt.title("Kiểm tra các giá trị ngoại lai cho toàn bộ đặc tính")
plt.show()
```
